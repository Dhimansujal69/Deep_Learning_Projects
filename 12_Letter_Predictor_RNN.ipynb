```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

import tensorflow as tf
from tensorflow import keras
```

```python
vocab = ["a","b","c","d"]
char_to_index = {ch: i for i, ch in enumerate(vocab)}

num_features = len(vocab)
```

```python
input_sequence = char_to_index['a'], char_to_index['b'], char_to_index['c']
target_sequence = char_to_index['d']
```

```python
x = tf.one_hot([input_sequence],depth=num_features)
y = tf.one_hot([target_sequence],depth=num_features)
```

```python
model = keras.Sequential([
    keras.layers.SimpleRNN(
        units=8,
        input_shape=(3, num_features),
        return_sequences=False
    ),
    keras.layers.Dense(units=num_features, activation='softmax')
])
```

```python
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(
    x,
    y,
    epochs=100,
    verbose=1
)
```

```python
predictions = model.predict(x)
```

```python
label_predictions = np.argmax(predictions,  axis=-1)
print(label_predictions)
```
