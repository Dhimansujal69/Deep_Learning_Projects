```python
import tensorflow as tf

tensor = tf.constant([
    [1,2,3],
    [4,5,6]
])

print(tensor)


print(tensor.shape)
```

# Tensor Indexing

```python
tensor_1 = tf.constant([
    [10,20,30],
    [40,50,60]
])

print(tensor_1)
```

```python
print(tensor_1[0,0])
```

```python
print(tensor_1[1,1])
```

```python
print(tensor_1[1])
```

```python
print(tensor_1[0,2])
```

# Tensor Slicing 

```python
tensor = tf.constant([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

print(tensor.shape)
```

```python
print(tensor)
```

```python
print(tensor[:2])
```

```python
print(tensor[1::2])
```

```python
print(tensor[: , :2])
```

```python
print(tensor[1::1])
```

```python
print(tensor[: , 2])
```

# Tensor Reshaping

```python
tensor = tf.constant([1,2,3,4,5,6])

reshaped = tf.reshape(tensor,(2,3))

print(reshaped)
```

```python
reshaped  =tf.reshape(tensor,(3,2))
print(reshaped)
```

# Expanding Dimensions 

```python
tensor = tf.constant([1,2,3])
print (tensor.shape)
expanded = tf.expand_dims(tensor,axis=0)

print(expanded)

print(expanded.shape)
```

```python
expanded = tf.expand_dims(tensor,axis=1)

print(expanded)
```

# Squeeze

```python
tensor = tf.constant([[1,2,3]])

print(tensor.shape)
```

```python
squeezed = tf.squeeze(tensor)

print(squeezed)
print(squeezed.shape)
```

# Changing DataType

```python
tensor = tf.constant([1,2,3])

print(tensor.dtype)
```

```python
new_tensor = tf.cast(tensor , tf.float32)

print(new_tensor.dtype)
```

# Creating Random Tensor

```python
random = tf.random.uniform(
    shape=(2,3),
    minval=0,
    maxval=10,
    dtype=tf.int32
)

print(random)
```

```python
# Tensor With Decimal Values 

random = tf.random.normal(shape=(2,3))

print(random)
```

# Tensor to Numpy

```python
tensor = tf.constant([1,2,3])

array = tensor.numpy()

print(array)

print(array.dtype)
```

```python
import numpy as np
```

```python
array = np.array([1,2,3])

tensor = tf.constant(array)

print(tensor)
```
