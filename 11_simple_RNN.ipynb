```python
import numpy as np

sequence = np.arange(1,101)

sequence[:10]
```

```python
x = []
y = []
```

```python
for i in range(len(sequence) -5):
    x.append(sequence[i:i+5])
    y.append(sequence[i+5])
```

```python
print(x[:3])
print(y[:3])
```

```python
x = np.array(x)
y = np.array(y)

print("Shape of X: ",x.shape)
print("Shape of Y: ",y.shape)
```

```python
x = x.reshape(95, 5, 1)
print("Shape of X: ",x.shape)
```

```python
model = keras.Sequential([

    SimpleRNN(32, input_shape=(5,1)),
    Dense(1)
])
```

```python
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)
```

```python
model.fit(
    x,
    y,
    epochs=100,
    verbose=1
)
```
