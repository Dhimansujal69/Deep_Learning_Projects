```python
import numpy as np

sequence = np.arange(1,101)

sequence[:10]
```

```python
x = []
y = []
```

```python
for i in range(len(sequence) -5):
    x.append(sequence[i:i+5])
    y.append(sequence[i+5])
```

```python
print(x[:3])
print(y[:3])
```

```python
x = np.array(x)
y = np.array(y)

print("Shape of X: ",x.shape)
print("Shape of Y: ",y.shape)
```

```python
x = x.reshape(95, 5, 1)
print("Shape of X: ",x.shape)
```

```python
model = keras.Sequential([

    SimpleRNN(32, input_shape=(5,1)),
    Dense(1)
])
```

```python
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)
```

```python
model.fit(
    x,
    y,
    epochs=100,
    verbose=1
)
```

```python
from sklearn.preprocessing import MinMaxScaler

scalar = MinMaxScaler()

sequence_scaled = scalar.fit_transform(sequence.reshape(-1,1))
```

```python
print(sequence_scaled[:5])
print(sequence_scaled[-5:])
```

```python
X_scaled = []
y_scaled = []

for i in range(len(sequence_scaled) - 5):
    X_scaled.append(sequence_scaled[i:i+5])
    y_scaled.append(sequence_scaled[i+5])

X_scaled = np.array(X_scaled)
y_scaled = np.array(y_scaled)
```

```python
print("X shape:", X_scaled.shape)
print("y shape:", y_scaled.shape)
```

```python
model_scaled = Sequential([
    SimpleRNN(32, input_shape=(5, 1)),
    Dense(1)
])

model_scaled.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)
```

```python
history_scaled = model_scaled.fit(
    X_scaled,
    y_scaled,
    epochs=100,
    verbose=1
)
```

```python
new_input = np.array([[96], [97], [98], [99], [100]])
```

```python
new_input_scaled = scalar.transform(new_input)

new_input_scaled = new_input_scaled.reshape(1,5,1)
```

```python
prediction_scaled = model_scaled.predict(new_input_scaled)

print(prediction_scaled)
```

```python
prediction_original = scalar.inverse_transform(prediction_scaled)

print(prediction_original)
```
