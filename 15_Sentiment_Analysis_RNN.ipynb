```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
```

```python
# Sample reviews
sentences = [
    "i loved this movie it was fantastic",
    "what an amazing and wonderful experience",
    "best film i have seen in a long time",
    "absolutely incredible acting and great plot",
    "i hated this movie it was terrible",
    "worst film ever completely waste of time",
    "awful acting and horrible script",
    "boring and completely disappointing experience"
]
```

```python
labels = np.array([1,1,1,1,0,0,0,0], dtype=np.float32)
```

```python
max_len = 10
vectorizer = tf.keras.layers.TextVectorization(output_mode = 'int', output_sequence_length = max_len)
```

```python
vectorizer.adapt(sentences)
```

```python
vocab_size = len(vectorizer.get_vocabulary())
```

```python
vocab_size
```

```python
x_train = vectorizer(np.array(sentences)).numpy()
y_train = labels
```

```python
print("Shape of X_train:", x_train.shape)
print("Vocabulary Size:", vocab_size)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(max_len,)),
    tf.keras.layers.Embedding(input_dim=vocab_size, output_dim=8, mask_zero=True),
    tf.keras.layers.LSTM(units=16, return_sequences=False),
    tf.keras.layers.Dense(units=1, activation='sigmoid')
])
```

```python
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(
    x_train,
    y_train,
    epochs=150,
    verbose=2
)
```

```python
test_reviews = np.array([
    "this movie was fantastic and wonderful",
    "terrible film complete waste"
])

test = vectorizer(np.array(test_reviews)).numpy()

pred = model.predict(test)

print(pred)
```
